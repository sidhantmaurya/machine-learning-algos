```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x= np.array([1, 2, 3, 4, 5], dtype=float)
y= np.array([2, 4, 5, 4, 5], dtype=float)
```

```python
x_mean=np.mean(x) #mean=4
y_mean=np.mean(y) # mean=13.28
```

```python
numenator=np.sum((x-x_mean)*(y-y_mean))
denomenator= np.sum((x-x_mean)**2)
```

```python
slope=numenator/denomenator
```

```python
intercept=y_mean-slope*x_mean
```

```python
def priduct(x,slope,intercept):
  return intercept + slope*x

x_priduct=6
y_pre=priduct(x_priduct,slope,intercept)

print(f"{x_priduct}  and {y_pre}")
```

```python
y_pridict=priduct(x,slope,intercept)

mse=np.mean((y-y_pridict)**2)


ss_num=np.sum((y-y_pridict)**2)
ss_deno=np.sum((y-y_mean)**2)

r=1-(ss_num/ss_deno)

print(f'mse {mse : .4f}')
print(f"r {r: .4f}")
```

```python
x_mean=np.mean(x)
y_mean=np.mean(y)

numen=np.sum((x-x_mean)*(y-y_mean))
deno=np.sum((x-x_mean)**2)

slope=numen/deno
slope
```

```python
intercepr=y_mean-slope*x_mean
intercepr
```

```python
def predict(x,slope,intercepr):
  return slope*x+intercepr

x_new=6
y_pridu=predict(x_new,slope,intercepr)
print(f"{x_new } predict is {y_pridu}")
```

```python
y_pridu=predict(x,slope,intercepr)

mse = np.sum((y-y_pridu)**2)
```
